```python
from sklearn.datasets import make_classification
import numpy as np
import matplotlib.pyplot as plt
```

```python
x , y = make_classification(n_samples=100,n_features=2,n_informative=1,n_redundant=0,
                            n_classes=2, n_clusters_per_class=1,random_state=42,hypercube=False,class_sep=15)
```

```python
x,y
```

```python
plt.figure(figsize=(10,6))
plt.scatter(x[:,1],x[:,0],c=y,cmap='winter',s=100)
```

```python
def perceptron(x,y):
    w1=w2=b=1
    lr = 0.1

    for j in range(1000):
        for i in range(x.shape[0]):
            z = w1 * x[i][0] + w2 * x[i][1]+b  # predicted point
            if z * y[i] < 0:
                w1 = w1 + lr * y[i] * x[i][0]
                w2 = w2 + lr * y[i] * x[i][0]
                b = b  + lr * y[i] 
    return w1 , w2 , b
```

```python
w1,w2,b =perceptron(x,y)
```

```python
m = -(w1/w2)
c = -(b/w2)
print(m,c)
```

```python
x_input = np.linspace(-3,3,100)
y_input = m * x_input * c

plt.figure(figsize=(10,6))
plt.plot(x_input,y_input,color='red',linewidth=3)
plt.scatter(x[:,1],x[:,0],c=y,cmap='winter',s=100)
plt.ylim(-3,2)
```
